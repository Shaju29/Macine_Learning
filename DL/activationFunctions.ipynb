```python
import math
```

Sigmoid activation Function

```python
def sigmoid(x):
    return (1/1+math.exp(-x))
```

```python
sigmoid(100)
```

tanh_activation Function


```python
def tanh(x):
    return ((math.exp(x)-math.exp(-x))/(math.exp(x)+math.exp(-x)))
```

```python
tanh(-56)
```

ReLU

```python
def ReLU(x):
    return max(0,x)
```

```python
ReLU(-100)
```

Leaky ReLU

```python
def leaky_relu(x):
    return max(0.1*x,x)
```

```python
leaky_relu(-100)
```
